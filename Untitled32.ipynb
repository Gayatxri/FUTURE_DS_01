# College Event Feedback Analysis
## Future Interns – Data Science & Analytics Task 3


```python
import pandas as pd

df = pd.read_csv("student_feedback.csv")
df.head()
```

```python
df.columns
```

```python
df.isnull().sum()
df.dropna(inplace=True)

```

```python
df.head()
df.columns
df.dtypes

```

```python
rating_cols = [
    "Well versed with the subject",
    "Explains concepts in an understandable way",
    "Use of presentations",
    "Degree of difficulty of assignments",
    "Solves doubts willingly",
    "Structuring of the course",
    "Provides support for students going above and beyond",
    "Course recommendation based on relevance"
]

df["Overall_Rating"] = df[rating_cols].mean(axis=1)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x=df["Overall_Rating"].round())
plt.title("Overall Student Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Students")
plt.show()
```

```python
sns.countplot(x="Explains concepts in an understandable way", data=df)
plt.title("Clarity of Explanation Rating")
plt.show()
```

## Key Insights
- The overall student satisfaction rating is generally high, indicating a positive response to the events.
- Most students rated aspects like subject knowledge and course structuring positively.
- Some variation was observed in ratings related to clarity of explanation and assignment difficulty.
- Support provided to students was well appreciated, showing strong instructor engagement.



## Recommendations
- Improve clarity of explanation by including more examples and interactive discussions.
- Balance assignment difficulty to match student expectations.
- Continue providing strong support and doubt-clearing sessions.
- Collect feedback regularly to further improve event quality.
